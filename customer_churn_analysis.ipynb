# Customer Churn Analysis
Exploratory analysis of a synthetic subscription-customer dataset using Python and Pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/customer_churn.csv')
df.head()
```

## Data Quality Checks

```python
print('Shape:', df.shape)
print('Duplicates:', df.duplicated().sum())
display(df.isna().sum().to_frame('missing_values'))
```

## Overall Churn Rate

```python
overall_churn = df['Churned'].mean() * 100
print(f'Overall churn rate: {overall_churn:.2f}%')
```

## Churn by Contract Type

```python
contract = df.groupby('ContractType')['Churned'].mean().mul(100).sort_values(ascending=False)
display(contract.to_frame('Churn Rate %'))
contract.plot(kind='bar', title='Churn Rate by Contract Type')
plt.ylabel('Churn Rate (%)')
plt.tight_layout()
plt.show()
```

## Churn by Tenure

```python
df['TenureBand'] = pd.cut(df['TenureMonths'], bins=[0,6,12,24,48,72], labels=['0-6','7-12','13-24','25-48','49-72'])
tenure = df.groupby('TenureBand', observed=False)['Churned'].mean().mul(100)
display(tenure.to_frame('Churn Rate %'))
tenure.plot(kind='bar', title='Churn Rate by Tenure Band')
plt.ylabel('Churn Rate (%)')
plt.tight_layout()
plt.show()
```

## Churn by Support Tickets

```python
df['TicketBand'] = pd.cut(df['SupportTickets'], bins=[-1,0,2,4,100], labels=['0','1-2','3-4','5+'])
tickets = df.groupby('TicketBand', observed=False)['Churned'].mean().mul(100)
display(tickets.to_frame('Churn Rate %'))
tickets.plot(kind='bar', title='Churn Rate by Support Ticket Band')
plt.ylabel('Churn Rate (%)')
plt.tight_layout()
plt.show()
```

## Churn by Satisfaction Score

```python
satisfaction = df.groupby('SatisfactionScore')['Churned'].mean().mul(100)
display(satisfaction.to_frame('Churn Rate %'))
satisfaction.plot(kind='bar', title='Churn Rate by Satisfaction Score')
plt.ylabel('Churn Rate (%)')
plt.tight_layout()
plt.show()
```
